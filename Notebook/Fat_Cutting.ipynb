Data import

```python
import pandas as pd 
```

```python
Sales           = pd.read_csv('CSVDATA/True sales.csv')
Purchase        = pd.read_csv('CSVDATA/PurchasesFINAL12312016.csv')
```

Sales table optimisation 

```python
Sales.info()
```

```python
Sales_MinMax = Sales.agg(['min', 'max','dtype']).T
Sales_MinMax
```

```python
type_map = {
    'InventoryId'       :'category',
    'Store'             :'int8',
    'Brand'             :'int32',
    'Description'       :'category',
    'Size'              :'category',
    'SalesQuantity'     :'int16',
    'SalesDollars'      :'float16', # i dont recommend float 16 but its at least safe for this analysis
    'SalesPrice'        :'float16', 
    'Volume'            :'int16',
    'Classification'    :'int8'  , 
    'ExciseTax'         :'float16',
    'VendorNo'          :'int32',
    'VendorName'        :'category'
}

Sales = Sales.astype(type_map)
```

```python
Sales['SalesDate']= pd.to_datetime(Sales['SalesDate']).dt.normalize()
```

```python
Sales.info()
```

Purchase data optimisation

```python
Purchase.info()
```

```python
Purchase.nunique()
```

```python
Purchase_MinMax = Purchase.agg(['min' , 'max','dtype']).T
Purchase_MinMax
```

```python
dtype_map = {
    'InventoryId'   :'category',
    'Store'         :'int8',
    'Brand'         :'int32',
    'Description'   :'category',
    'Size'          :'category',
    'VendorNumber'  :'int32',
    'VendorName'    :'category',
    'PONumber'      :'int16',
    'PurchasePrice' :'float16',
    'Quantity'      :'int16',
    'Dollars'       :'float32',
    'Classification':'int8'	
}
Purchase=Purchase.astype(dtype_map)
```

```python
Purchase[['PODate','ReceivingDate','InvoiceDate','PayDate']]=Purchase[['PODate','ReceivingDate','InvoiceDate','PayDate']].apply(pd.to_datetime).apply(lambda x : x.dt.normalize())
```
